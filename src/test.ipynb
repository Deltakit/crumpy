```python
from __future__ import annotations

stim_circuit = """QUBIT_COORDS(1, 0) 0
QUBIT_COORDS(2, 0) 1
QUBIT_COORDS(3, 0) 2
CX 1 2
MARKX(0) 0
TICK
CX 0 2
TICK
H 0
SWAP 0 1
"""
```

```python
import anywidget
import traitlets


class CircuitWidget(anywidget.AnyWidget):
    _esm = "./crumpy/bundle.js"
    stim = traitlets.Unicode("").tag(sync=True)
    indentCircuitLines = traitlets.Bool(True).tag(sync=True)
    curveConnectors = traitlets.Bool(True).tag(sync=True)


circuit = CircuitWidget(stim=stim_circuit)
circuit
```

```python
circuit.curveConnectors = False
circuit.indentCircuitLines = False
```

```python
circuit.stim = """QUBIT_COORDS(0, 0) 0
QUBIT_COORDS(1, 0) 1
QUBIT_COORDS(100, 0) 2
X 0
TICK
Y 1
TICK
Z 2
"""
```

```python
import ipywidgets


def sliderToCircuit(slider_val):
    base_circuit = """QUBIT_COORDS(0, 0) 0
    QUBIT_COORDS(1, 0) 1
    #!pragma MARKX(0) 0
    TICK
    CX 0 1
    """
    h_count = int(slider_val)
    return base_circuit + "H 0\n" * h_count


circuit2 = CircuitWidget()
slider = ipywidgets.IntSlider(description="# of H gates", value=1, min=0, max=10)
ipywidgets.dlink((slider, "value"), (circuit2, "stim"), sliderToCircuit)
display(slider, circuit2)
```
